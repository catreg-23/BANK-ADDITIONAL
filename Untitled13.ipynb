```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

LOAD DATA AND PREPROCESS

```python
file_path='twitter_training.csv'
columns=['Tweet_ID','Entity','Sentiment','Tweet_Text']
df=pd.read_csv(file_path,names=columns)
```

DROP MISSING VALUES

```python
df.dropna(subset=['Tweet_Text'],inplace=True)
```

SET VISUAL STYLE

```python
sns.set_theme(style="whitegrid")
```

PLOT OVERALL SENTIMENT DISTRIBUTION

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df,x='Sentiment',palette='mako',order=['Positive','Negative','Neutral','Irrelevent'])
plt.title('Overall Sentiment Distribution Across All Social Media Posts',fontsize=14)
plt.xlabel('Tweet Count',fontsize=11)
plt.tight_layout()
plt.show()
```

PLOT BRAND/ENTITY-SPECIFIC SENTIMENT

```python
top_entities=df['Entity'].value_counts().head(5).index
df_top=df[df['Entity'].isin(top_entities)]
plt.figure(figsize=(12,6))
sns.countplot(data=df_top,x='Entity',hue='Sentiment',palette=['purple','pink','grey','blue'],hue_order=['Positive','Negative','Neutral','Irrelevent'])
plt.title('Sentiment Breakdown Across Top 5 Brands/Entities',fontsize=14)
plt.xlabel('Brand/Topic',fontsize=11)
plt.ylabel('Tweet Counr',fontsize=11)
plt.legend(title='Sentiment')
plt.xticks(rotation=25)
plt.tight_layout()
plt.show()
```

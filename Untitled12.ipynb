```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier,plot_tree
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
```

LOAD THE DATASET

```python
df=pd.read_csv('bank-additional-full.csv',sep=';')
```

PREPROCESSING AND ENCODING


```python
df['y']=df['y'].map({'yes':1,'no':0})
x=pd.get_dummies(df.drop(columns=['y']),drop_first=True)
y=df['y']
```

TRAIN-TEST SPLIT(80% TRAIN,20% TEST)

```python
X_train,X_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42,stratify=y)
```

INITIALIZE AND TRAIN THEDECISION TREE MODEL

```python
clf = DecisionTreeClassifier(criterion='gini',max_depth=5,random_state=42)
clf.fit(X_train,y_train)
```

MODEL EVALUATION

```python
y_pred=clf.predict(X_test)
print(f"Model Accuracy:{accuracy_score(y_test,y_pred)*100:.2f}%\n")
print("Confusion Matrix:")
print(confusion_matrix(y_test,y_pred))
print("\nClassification Report:")
print(classification_report(y_test,y_pred,target_names=['No Purchase','Purchase']))
```

TREE VISUALIZATION

```python
plt.figure(figsize=(24,12))
plot_tree(clf,feature_names=x.columns,class_names=['No Purchase','Purchase'],filled=True,fontsize=9)
plt.title("Decision Tree Classifier-Customer Purchase Prdiction")
plt.tight_layout()
plt.savefig("decision_tree_output.png")
plt.show()
```
